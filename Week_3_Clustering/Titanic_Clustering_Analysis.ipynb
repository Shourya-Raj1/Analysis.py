# Week 3 – Unsupervised Learning and Clustering Analysis
Titanic Cleaned Dataset

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```
df=pd.read_csv('titanic_cleaned.csv')
df.head()
```

## Feature Selection

```
features=df[['Age','Fare','FamilySize']]
scaler=StandardScaler()
scaled=scaler.fit_transform(features)
```

## Apply K-Means Clustering

```
kmeans=KMeans(n_clusters=3, random_state=42, n_init=10)
df['Cluster']=kmeans.fit_predict(scaled)
df[['Age','Fare','FamilySize','Cluster']].head()
```

## Cluster Visualization

```
plt.figure(figsize=(6,4))
sns.scatterplot(data=df,x='Age',y='Fare',hue='Cluster',palette='Set2')
plt.title('K-Means Clusters: Age vs Fare')
plt.show()
```

## Cluster Distribution

```
plt.figure(figsize=(5,4))
sns.countplot(data=df,x='Cluster',palette='Set2')
plt.title('Passengers in Each Cluster')
plt.show()
```

## Cluster Characteristics

```
df.groupby('Cluster')[['Age','Fare','FamilySize']].mean()
```
